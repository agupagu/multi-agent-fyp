```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick

# Price per input token
claude_input_price = 0.000003
openai_input_price = 0.0000025
```

# Simple Analysis of Data from Different Agents


## Layer3 LoginChecker Analysis (Claude)


```python
Layer3_LoginChecker_Claude_df = pd.read_excel('FYP Data.xlsx', sheet_name='Layer3 LoginChecker Claude')
Layer3_LoginChecker_Claude_df['Cost'] = Layer3_LoginChecker_Claude_df['Tokens'] * claude_input_price
Layer3_LoginChecker_Claude_df.describe()
```

## Layer3 LoginChecker Analysis (OpenAI)


```python
Layer3_LoginChecker_OpenAI_df = pd.read_excel('FYP Data.xlsx', sheet_name='Layer3 LoginChecker OpenAI')
Layer3_LoginChecker_OpenAI_df['Cost'] = Layer3_LoginChecker_OpenAI_df['Tokens'] * openai_input_price
Layer3_LoginChecker_OpenAI_df.describe()
```

```python
# Calculate average cost and standard deviation for each agent
avg_cost_claude = Layer3_LoginChecker_Claude_df['Cost'].mean()
std_cost_claude = Layer3_LoginChecker_Claude_df['Cost'].std()
avg_cost_openai = Layer3_LoginChecker_OpenAI_df['Cost'].mean()
std_cost_openai = Layer3_LoginChecker_OpenAI_df['Cost'].std()

# Prepare data for plotting
agents = ['Claude', 'OpenAI']
avg_costs = [avg_cost_claude, avg_cost_openai]
std_costs = [std_cost_claude, std_cost_openai]

# Create a bar chart with error bars representing the standard deviation
plt.figure(figsize=(8,6))
plt.bar(agents, avg_costs, yerr=std_costs, capsize=10, color=['salmon', 'skyblue'])
plt.ylabel('Average Cost')
plt.title('Comparison of Average Cost with Standard Deviation (LoginChecker Layer3)')
# Format the y-axis to display two decimal places
ax = plt.gca()
ax.yaxis.set_major_formatter(mtick.FormatStrFormatter('%.3f'))
plt.show()
```

## Galxe OtherAgents (Claude)

### This analysis is meant to show the difference in task success rate for Claude and OpenAI (The models used are 3.7 Sonnet and GPT-4o)

```python
Galxe_OtherAgents_Claude_df = pd.read_excel('FYP Data.xlsx', sheet_name='Galxe OtherAgents Claude')

#Calculate price
Galxe_OtherAgents_Claude_df['TaskCompletionAgent_Price'] = Galxe_OtherAgents_Claude_df['TaskCompletionAgent_Tokens'] * claude_input_price
Galxe_OtherAgents_Claude_df['TaskVerificationAgent_Price'] = Galxe_OtherAgents_Claude_df['TaskVerificationAgent_Tokens'] * claude_input_price
Galxe_OtherAgents_Claude_df['QuestCompletionAgent_Price'] =   Galxe_OtherAgents_Claude_df['QuestCompletionAgent_Tokens'] * claude_input_price


Galxe_OtherAgents_Claude_df['Total_Price'] = (
    Galxe_OtherAgents_Claude_df['TaskCompletionAgent_Price'] +
    Galxe_OtherAgents_Claude_df['TaskVerificationAgent_Price'] +
    Galxe_OtherAgents_Claude_df['QuestCompletionAgent_Price']
)


# Average task completion rate

Galxe_total_tasks = Galxe_OtherAgents_Claude_df['Total Tasks'].sum()
Galxe_tasks_completed = Galxe_OtherAgents_Claude_df['Tasks Completed'].sum()
average_task_completion_rate_claude = (Galxe_tasks_completed / Galxe_total_tasks) * 100

Galxe_OtherAgents_Claude_df.describe()
print(f"The Average Task Completion Rate is {average_task_completion_rate_claude:.2f}%")
```

## Galxe OtherAgents (OpenAI)

```python
Galxe_OtherAgents_OpenAI_df = pd.read_excel('FYP Data.xlsx', sheet_name='Galxe OtherAgents OpenAI')

# Average task completion rate

Galxe_total_tasks = Galxe_OtherAgents_OpenAI_df['Total Tasks'].sum()
Galxe_tasks_completed = Galxe_OtherAgents_OpenAI_df['Tasks Completed'].sum()
average_task_completion_rate_openai = (Galxe_tasks_completed / Galxe_total_tasks) * 100

Galxe_OtherAgents_OpenAI_df.describe()
print(f"The Average Task Completion Rate is {average_task_completion_rate_openai:.2f}%")
```

```python
# Prepare data for plotting
agents = ['Claude', 'OpenAI']
completion_rates = [average_task_completion_rate_claude, average_task_completion_rate_openai]

# Create the bar chart
plt.figure(figsize=(8,6))
bars = plt.bar(agents, completion_rates, color=['salmon', 'skyblue'])

# Adding the values on top of the bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 1, f'{yval:.2f}%', ha='center', va='bottom')

plt.ylabel('Average Task Completion Rate (%)')
plt.title('Average Task Completion Rate Comparison (Galxe)')
plt.ylim(0, 100)  # Adjust if your rates exceed 100%
plt.show()
```
