```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

claude_input_price = 0.000003
openai_input_price = 0.0000025
```

# Analysis of Data from Different Agents


## Layer3 LoginChecker Analysis (Claude)


```python
Layer3_LoginChecker_Claude_df = pd.read_excel('FYP Data.xlsx', sheet_name='Layer3 LoginChecker Claude')
Layer3_LoginChecker_Claude_df.describe()
```

```python

# Set the style (optional)
plt.style.use('dark_background') 

# Create the scatterplot
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Time', y='Tokens', data=Layer3_LoginChecker_Claude_df)

# Add a regression line to show trend 
sns.regplot(x='Time', y='Tokens', data=Layer3_LoginChecker_Claude_df, 
            scatter=False, line_kws={"color":"orange"})

# Add labels and title
plt.title('Tokens vs Time for Layer3 LoginChecker Agent (Claude)', fontsize=16)
plt.xlabel('Time (seconds)', fontsize=12)
plt.ylabel('Tokens Used', fontsize=12)

# Add correlation coefficient as text 
corr = Layer3_LoginChecker_Claude_df['Tokens'].corr(Layer3_LoginChecker_Claude_df['Time'])
plt.annotate(f'Correlation: {corr:.2f}', xy=(0.7, 0.05), xycoords='axes fraction', 
             fontsize=12, bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="black", alpha=0.7))

# Show grid 
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Layer3 OtherAgents Claude

```python
Layer3_OtherAgents_Claude_df = pd.read_excel('FYP Data.xlsx', sheet_name='Layer3 OtherAgents Claude')

#Calculate price
Layer3_OtherAgents_Claude_df['TaskCompletionAgent_Price'] = Layer3_OtherAgents_Claude_df['TaskCompletionAgent_Tokens'] * claude_input_price
Layer3_OtherAgents_Claude_df['TaskVerificationAgent_Price'] = Layer3_OtherAgents_Claude_df['TaskVerificationAgent_Tokens'] * claude_input_price
Layer3_OtherAgents_Claude_df['QuestCompletionAgent_Price'] =   Layer3_OtherAgents_Claude_df['QuestCompletionAgent_Tokens'] * claude_input_price


Layer3_OtherAgents_Claude_df['Total_Price'] = (
    Layer3_OtherAgents_Claude_df['TaskCompletionAgent_Price'] +
    Layer3_OtherAgents_Claude_df['TaskVerificationAgent_Price'] +
    Layer3_OtherAgents_Claude_df['QuestCompletionAgent_Price']
)


Layer3_OtherAgents_Claude_df.describe()
```
