```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick

# Price per input token
claude_input_price = 0.000003
openai_input_price = 0.0000025
```

# Simple Analysis of Data from Different Agents


## Layer3 LoginChecker Analysis (Claude)


```python
Layer3_LoginChecker_Claude_df = pd.read_excel('FYP Data.xlsx', sheet_name='Layer3 LoginChecker Claude')
Layer3_LoginChecker_Claude_df['Cost'] = Layer3_LoginChecker_Claude_df['Tokens'] * claude_input_price
Layer3_LoginChecker_Claude_df.describe()
```

## Layer3 LoginChecker Analysis (OpenAI)


```python
Layer3_LoginChecker_OpenAI_df = pd.read_excel('FYP Data.xlsx', sheet_name='Layer3 LoginChecker OpenAI')
Layer3_LoginChecker_OpenAI_df['Cost'] = Layer3_LoginChecker_OpenAI_df['Tokens'] * openai_input_price
Layer3_LoginChecker_OpenAI_df.describe()
```

```python
import matplotlib.ticker as mtick

# Calculate average cost and standard deviation for each agent
avg_cost_claude = Layer3_LoginChecker_Claude_df['Cost'].mean()
std_cost_claude = Layer3_LoginChecker_Claude_df['Cost'].std()
avg_cost_openai = Layer3_LoginChecker_OpenAI_df['Cost'].mean()
std_cost_openai = Layer3_LoginChecker_OpenAI_df['Cost'].std()

print(f"The average cost for Claude is {avg_cost_claude:.3f} with a standard deviation of {std_cost_claude:.3f}")
print(f"The average cost for OpenAI is {avg_cost_openai:.3f} with a standard deviation of {std_cost_openai:.3f}")

# Prepare data for plotting
agents = ['Claude', 'OpenAI']
avg_costs = [avg_cost_claude, avg_cost_openai]
std_costs = [std_cost_claude, std_cost_openai]

# Create a bar chart with black background and white error bars
plt.figure(figsize=(8,6), facecolor='black')
ax = plt.gca()
ax.set_facecolor('black')

# Set spines, ticks, labels, and title to white for visibility
for spine in ax.spines.values():
    spine.set_color('white')
ax.tick_params(colors='white')
ax.yaxis.label.set_color('white')
ax.xaxis.label.set_color('white')
ax.title.set_color('white')

# Plot the bars with white error bars
bars = plt.bar(agents, avg_costs, yerr=std_costs, capsize=10, color=['salmon', 'skyblue'], ecolor='white')

# Format the y-axis to display up to three decimal places
ax.yaxis.set_major_formatter(mtick.FormatStrFormatter('%.3f'))

# Add cost ± std inside each bar with black text
for bar, avg, std in zip(bars, avg_costs, std_costs):
    height = bar.get_height()
    text = f'{avg:.3f} \u00B1 {std:.3f}'
    plt.text(bar.get_x() + bar.get_width()/2, height * 0.5, text,
             ha='center', va='center', color='black', fontsize=10, fontweight='bold')

plt.ylabel('Average Cost')
plt.title('Comparison of Average Cost with Standard Deviation (LoginChecker Layer3)')

plt.show()
```

## Comparison between Claude and User for time taken on Layer3 Quests

```python
Layer3_users_df = pd.read_excel('FYP Data.xlsx', sheet_name='Layer3 User')
Galxe_users_df = pd.read_excel('FYP Data.xlsx', sheet_name='Galxe User')
```

```python
import matplotlib.pyplot as plt

# Ensure quest names are strings
Layer3_users_df['Quest Name'] = Layer3_users_df['Quest Name'].astype(str)
Layer3_OtherAgents_Claude_df['Quest Name'] = Layer3_OtherAgents_Claude_df['Quest Name'].astype(str)

# Sort for consistent alignment
Layer3_users_df = Layer3_users_df.sort_values(by='Quest Name')
Layer3_OtherAgents_Claude_df = Layer3_OtherAgents_Claude_df.sort_values(by='Quest Name')

# Create plot
plt.figure(figsize=(12,6), facecolor='black')
ax = plt.gca()
ax.set_facecolor('black')

# Plot lines
plt.plot(Layer3_users_df['Quest Name'], Layer3_users_df['Average Total Time'], 
         label='Users', marker='o', linestyle='-', color='salmon')
plt.plot(Layer3_OtherAgents_Claude_df['Quest Name'], Layer3_OtherAgents_Claude_df['Total Time'], 
         label='Claude Agent', marker='s', linestyle='--', color='skyblue')

# Formatting for dark theme
for spine in ax.spines.values():
    spine.set_color('white')
ax.tick_params(colors='white')
ax.yaxis.label.set_color('white')
ax.xaxis.label.set_color('white')
ax.title.set_color('white')

# White grid lines
plt.grid(True, linestyle='--', linewidth=0.5, color='white', alpha=0.3)

# Labels and layout
plt.xlabel('Quest Name')
plt.ylabel('Average Total Time (seconds)')
plt.title('Comparison of Quest Completion Time: Users vs Claude Agent (Layer3)')
plt.xticks(rotation=45, ha='right', color='white')
plt.yticks(color='white')
plt.legend(facecolor='black', edgecolor='white', labelcolor='white')
plt.tight_layout()

plt.show()
```
